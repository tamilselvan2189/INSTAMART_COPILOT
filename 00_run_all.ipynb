## Final Verification: Check All Output Files

```python
print("\n" + "="*70)
print("FINAL VERIFICATION")
print("="*70)

required_files = {
    'Part A': [
        'data/clean/customers.csv',
        'data/clean/orders.csv',
        'data/clean/products.csv',
        'data/instamart.db',
        'sql/queries.sql',
    ],
    'Part B': [
        'outputs/customer_segments.csv',
        'outputs/retention_list.csv',
        'outputs/lapse_model.joblib',
        'outputs/lapse_logreg.joblib',
        'outputs/lapse_rf.joblib',
        'outputs/lapse_preprocessor.joblib',
    ],
    'Part C': [
        'outputs/triage_results.csv',
        'outputs/triage_budget_demo.csv',
    ],
    'Part D': [
        'outputs/agent_runs.jsonl',
        'outputs/refund_ledger.jsonl',
    ]
}

total_files = 0
found_files = 0

for part, files in required_files.items():
    print(f"\n{part}:")
    for file in files:
        total_files += 1
        file_path = Path(file)
        if file_path.exists():
            size = file_path.stat().st_size / 1024  # KB
            print(f"  ✓ {file} ({size:.1f} KB)")
            found_files += 1
        else:
            print(f"  ✗ {file} (MISSING)")

print(f"\n" + "="*70)
print(f"RESULT: {found_files}/{total_files} files found")
print(f"="*70)

if found_files == total_files:
    print("\n🎉 PROJECT EXECUTION COMPLETE! ALL FILES GENERATED.")
    print("\nThe project is ready for submission.")
else:
    print(f"\n⚠️  {total_files - found_files} file(s) missing. Check errors above.")
```
